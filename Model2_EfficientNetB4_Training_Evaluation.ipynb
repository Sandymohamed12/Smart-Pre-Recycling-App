# Model 2: Waste Material Classification using EfficientNetB4

This notebook documents the implementation, training, and evaluation of the waste material classification model.

The model classifies waste images into nine classes: E-waste, automobile waste, battery waste, glass waste, light bulbs, metal waste, organic waste, paper waste, and plastic waste.

dataset distribution table

```python
train_path = "data/wastes/train/"
test_path = "data/wastes/test/"

classes = sorted(os.listdir(train_path))

train_counts = []
test_counts = []

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

for cls in classes:
    train_class_path = os.path.join(train_path, cls)
    test_class_path = os.path.join(test_path, cls)

    train_count = len([
        f for f in os.listdir(train_class_path)
        if f.lower().endswith(image_extensions)
    ])

    test_count = len([
        f for f in os.listdir(test_class_path)
        if f.lower().endswith(image_extensions)
    ])

    train_counts.append(train_count)
    test_counts.append(test_count)

df_counts = pd.DataFrame({
    "Class": classes,
    "Train Images": train_counts,
    "Test Images": test_counts,
    "Total Images": np.array(train_counts) + np.array(test_counts)
})

df_counts
```

```python
x = np.arange(len(classes))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(x - width/2, train_counts, width, label="Train")
plt.bar(x + width/2, test_counts, width, label="Test")

plt.xlabel("Waste Class", fontsize=12)
plt.ylabel("Number of Images", fontsize=12)
plt.title("Train and Test Images per Waste Class", fontsize=14)

plt.xticks(x, classes, rotation=35, ha="right")
plt.grid(axis="y", alpha=0.25)
plt.legend(fontsize=10)

plt.tight_layout()
plt.savefig("train_test_images_per_class.png", dpi=300, bbox_inches="tight")
plt.show()
```

load dataset

```python
train_ds = tf.keras.preprocessing.image_dataset_from_directory(
    "data/wastes/train/",
    image_size=(380, 380),
    batch_size=32
)

val_ds = tf.keras.preprocessing.image_dataset_from_directory(
    "data/wastes/test/",
    image_size=(380, 380),
    batch_size=32
)
```

```python
class_names = train_ds.class_names
print("Classes:", class_names)

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
```

data augmentation

```python
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.2),
    layers.RandomContrast(0.2),
    layers.RandomZoom(0.2),
])
```

build EfficientNetB4 model

```python
base_model = tf.keras.applications.EfficientNetB4(
    include_top=False,
    input_shape=(380, 380, 3),
    weights="imagenet"
)

base_model.trainable = False

inputs = layers.Input(shape=(380, 380, 3))
x = data_augmentation(inputs)
x = tf.keras.applications.efficientnet.preprocess_input(x)

x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(len(class_names), activation="softmax")(x)

model = tf.keras.Model(inputs, outputs)
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

callbacks

```python
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_accuracy",
        patience=4,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        factor=0.2,
        patience=2,
        min_lr=1e-6
    ),
    tf.keras.callbacks.ModelCheckpoint(
        "best_effnetb4.h5",
        monitor="val_accuracy",
        save_best_only=True
    )
]
```

stage 1 training: frozen backbone

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=callbacks
)
```

stage 2 training: fine-tuning

```python
base_model.trainable = True
for layer in base_model.layers[:350]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history_finetune = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=callbacks
)
```

accuracy plot

```python
acc = history.history["accuracy"] + history_finetune.history["accuracy"]
val_acc = history.history["val_accuracy"] + history_finetune.history["val_accuracy"]

epochs = np.arange(1, len(acc) + 1)
fine_tuning_start = len(history.history["accuracy"]) + 1
x_ticks = [1, 5, 10, 15, 20, len(acc)]

plt.figure(figsize=(10, 5.5))

plt.plot(epochs, acc, linewidth=2.5, label="Training Accuracy")
plt.plot(epochs, val_acc, linewidth=2.5, label="Validation Accuracy")

plt.axvline(
    x=fine_tuning_start,
    linestyle="--",
    linewidth=2,
    color="gray",
    label="Fine-tuning Start"
)

plt.text(
    fine_tuning_start + 0.4,
    0.62,
    "Fine-tuning",
    fontsize=11,
    color="gray"
)

plt.xlabel("Epoch", fontsize=12)
plt.ylabel("Accuracy", fontsize=12)
plt.title("Training and Validation Accuracy Across Two Training Stages", fontsize=14)

plt.ylim(0.55, 0.95)
plt.xticks(x_ticks)

plt.grid(True, alpha=0.25, linewidth=0.8)
plt.legend(fontsize=10)
plt.tight_layout()

plt.savefig("final_accuracy_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

loss plot

```python
loss = history.history["loss"] + history_finetune.history["loss"]
val_loss = history.history["val_loss"] + history_finetune.history["val_loss"]

epochs = np.arange(1, len(loss) + 1)
fine_tuning_start = len(history.history["loss"]) + 1
x_ticks = [1, 5, 10, 15, 20, len(loss)]

plt.figure(figsize=(10, 5.5))

plt.plot(epochs, loss, linewidth=2.5, label="Training Loss")
plt.plot(epochs, val_loss, linewidth=2.5, label="Validation Loss")

plt.axvline(
    x=fine_tuning_start,
    linestyle="--",
    linewidth=2,
    color="gray",
    label="Fine-tuning Start"
)

plt.text(
    fine_tuning_start + 0.4,
    0.95,
    "Fine-tuning",
    fontsize=11,
    color="gray"
)

plt.xlabel("Epoch", fontsize=12)
plt.ylabel("Loss", fontsize=12)
plt.title("Training and Validation Loss Across Two Training Stages", fontsize=14)

plt.xticks(x_ticks)

plt.grid(True, alpha=0.25, linewidth=0.8)
plt.legend(fontsize=10)
plt.tight_layout()

plt.savefig("final_loss_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

model evaluation

```python
class_names = ['E-waste', 'automobile wastes', 'battery waste', 'glass waste',
               'light bulbs', 'metal waste', 'organic waste', 'paper waste', 'plastic waste']

test_loss, test_acc = model.evaluate(val_ds)
print("Test Loss:", test_loss)
print("Test Accuracy:", test_acc)
```

predictions for report and confusion matrix

```python
y_true = []
y_pred = []

for images, labels in val_ds:
    preds = model.predict(images)
    y_pred.extend(np.argmax(preds, axis=1))
    y_true.extend(labels.numpy())

print("\nClassification Report:\n")
print(classification_report(y_true, y_pred, target_names=class_names))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10,8))
sns.heatmap(cm, annot=True, fmt="d", xticklabels=class_names, yticklabels=class_names, cmap="Blues")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.show()
```

precision / recall / f1 plot

```python
report = classification_report(
    y_true,
    y_pred,
    target_names=class_names,
    output_dict=True
)

class_names_short = [
    "E-waste",
    "Automobile",
    "Battery",
    "Glass",
    "Light bulbs",
    "Metal",
    "Organic",
    "Paper",
    "Plastic"
]

precision = [report[name]["precision"] for name in class_names]
recall = [report[name]["recall"] for name in class_names]
f1 = [report[name]["f1-score"] for name in class_names]

x = np.arange(len(class_names_short))
width = 0.25

plt.figure(figsize=(12, 6))

plt.bar(x - width, precision, width, label="Precision")
plt.bar(x, recall, width, label="Recall")
plt.bar(x + width, f1, width, label="F1-score")

plt.xlabel("Waste Class", fontsize=12)
plt.ylabel("Score", fontsize=12)
plt.title("Precision, Recall, and F1-score per Waste Class", fontsize=14)

plt.xticks(x, class_names_short, rotation=35, ha="right")
plt.ylim(0.75, 1.0)

plt.grid(axis="y", alpha=0.25)
plt.legend(fontsize=10)
plt.tight_layout()

plt.savefig("precision_recall_f1_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

f1 per class

```python
plt.figure(figsize=(9, 5.5))

plt.barh(class_names_short, f1)

plt.xlabel("F1-score", fontsize=12)
plt.ylabel("Waste Class", fontsize=12)
plt.title("F1-score per Waste Class", fontsize=14)

plt.xlim(0.80, 1.00)
plt.grid(axis="x", alpha=0.25)

plt.tight_layout()
plt.savefig("f1_score_per_class.png", dpi=300, bbox_inches="tight")
plt.show()
```

final accuracy summary

```python
summary_names = ["Training Accuracy", "Validation Accuracy", "Test Accuracy"]

summary_values = [
    history_finetune.history["accuracy"][-1] * 100,
    history_finetune.history["val_accuracy"][-1] * 100,
    test_acc * 100
]

plt.figure(figsize=(8, 5))

plt.bar(summary_names, summary_values)

plt.ylabel("Accuracy (%)", fontsize=12)
plt.title("Final Model Accuracy Summary", fontsize=14)
plt.ylim(0, 100)

for i, v in enumerate(summary_values):
    plt.text(i, v + 1, f"{v:.2f}%", ha="center", fontsize=11)

plt.grid(axis="y", alpha=0.25)
plt.tight_layout()

plt.savefig("final_accuracy_summary.png", dpi=300, bbox_inches="tight")
plt.show()
```
